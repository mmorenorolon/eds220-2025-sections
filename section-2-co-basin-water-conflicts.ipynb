### Dataset Overview: Coded Water Conflict and Crisis Events in the Colorado River Basin (2005–2021)

This dataset documents water-related conflict and crisis events in the Upper Colorado River Basin, bounded by coordinates West: −119.35547, East: −103.71094, North: 31.65338, and South: 41.77131. Observations were collected between January 1, 2005 and December 31, 2021. The authors regard the dataset as a valuable resource for identifying conflict hotspots and analyzing social vulnerability and institutional resilience, contributing to ongoing USGS research and future water availability assessments.

**Citation**: Holloman, D.V., Hines, M.K., Zoanni, D.K. (2023). *Coded Water Conflict and Crisis Events in the Colorado River Basin, Derived from LexisNexis search 2005–2021*. U.S. Geological Survey.  
**Date of Access**: October 9, 2025  
**Archive Link**: [ScienceBase Catalog Entry](https://www.sciencebase.gov/catalog/item/63acac09d34e92aad3ca1480)
